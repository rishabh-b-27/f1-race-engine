```python
events = session.race_control_messages.copy()

events = events[
    events["Message"].str.contains(
        "VIRTUAL SAFETY CAR|SAFETY CAR",
        case=False,
        na=False
    )
][[
    "Time",
    "Message",
    "Lap"
]].copy()

events
```

```python
def classify_event(message):
    message = message.upper()

    if "VIRTUAL SAFETY CAR DEPLOYED" in message:
        return "VSC_START"
    
    if "VIRTUAL SAFETY CAR ENDING" in message:
        return "VSC_END"
    
    if "SAFETY CAR DEPLOYED" in message:
        return "SC_START"
    
    if "SAFETY CAR IN THIS LAP" in message:
        return "SC_END"
    
    return "OTHER"
```

```python
events["EventType"] = events["Message"].apply(classify_event)

events
```

```python
events = events[events["EventType"] != "OTHER"].copy()
```

```python
events[["Time", "Lap", "EventType"]].reset_index(drop=True)
```
